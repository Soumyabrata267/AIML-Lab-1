```python
from collections import deque

def solve_water_jug(capacity_a=4, capacity_b=3, target=2):
    # BFS Queue stores tuples of (current_state, path_to_this_state)
    # Each state is represented as (jug_a, jug_b)
    queue = deque([((0, 0), [])])
    visited = set([(0, 0)])

    while queue:
        (a, b), path = queue.popleft()

        # Check if we have reached the target in either jug
        if a == target or b == target:
            return path + [(a, b)]

        # Generate possible next states
        # 1. Fill Jug A
        # 2. Fill Jug B
        # 3. Empty Jug A
        # 4. Empty Jug B
        # 5. Pour A into B
        # 6. Pour B into A

        pour_a_to_b = min(a, capacity_b - b)
        pour_b_to_a = min(b, capacity_a - a)

        next_states = [
            ((capacity_a, b), "Fill Jug A"),
            ((a, capacity_b), "Fill Jug B"),
            ((0, b), "Empty Jug A"),
            ((a, 0), "Empty Jug B"),
            ((a - pour_a_to_b, b + pour_a_to_b), "Pour Jug A -> Jug B"),
            ((a + pour_b_to_a, b - pour_b_to_a), "Pour Jug B -> Jug A")
        ]

        for next_state, action in next_states:
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [(a, b, action)]))

    return None

# Solve the problem
steps = solve_water_jug()

if steps:
    print(f"Minimum sequence of steps to measure exactly 2 litres:")
    for i, step in enumerate(steps[:-1]):
        print(f"Step {i+1}: {step[2]:<22} -> Left with Jug A: {steps[i+1][0]}L, Jug B: {steps[i+1][1]}L")
else:
    print("No solution possible.")
```
