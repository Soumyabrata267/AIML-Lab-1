```python
import matplotlib.pyplot as plt
from collections import deque

# Adjacency list representation
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}


# BFS Traversal
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    result = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        result.append(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

    return result


# DFS Traversal
def dfs(graph, node, visited=None, result=None):
    if visited is None:
        visited = set()

    if result is None:
        result = []

    visited.add(node)
    result.append(node)

    for neighbour in graph[node]:
        if neighbour not in visited:
            dfs(graph, neighbour, visited, result)

    return result


# Draw tree using only Matplotlib
def draw_tree(graph):
    positions = {
        'A': (0, 3),
        'B': (-2, 2),
        'C': (2, 2),
        'D': (-3, 1),
        'E': (-1, 1),
        'F': (2, 1)
    }

    plt.figure(figsize=(8, 5))

    # Draw edges
    for node in graph:
        for neighbour in graph[node]:
            x1, y1 = positions[node]
            x2, y2 = positions[neighbour]

            plt.plot(
                [x1, x2],
                [y1, y2],
                color="black",
                linewidth=2
            )

    # Draw nodes
    for node, (x, y) in positions.items():
        plt.scatter(
            x, y,
            s=1500,
            color="skyblue",
            edgecolors="black",
            zorder=2
        )

        plt.text(
            x, y,
            node,
            ha="center",
            va="center",
            fontsize=14,
            fontweight="bold",
            zorder=3
        )

    plt.title("Graph / Tree Visualization")
    plt.axis("off")
    plt.show()


# Perform traversals
bfs_order = bfs(graph, 'A')
dfs_order = dfs(graph, 'A')

print("BFS Traversal:", " -> ".join(bfs_order))
print("DFS Traversal:", " -> ".join(dfs_order))

# Display tree
draw_tree(graph)
```
