```python
class RoadNetwork:
    def __init__(self):
        # Use an adjacency list to represent the graph
        self.adj_list = {}

    def add_road(self, city1, city2):
        """Adds an undirected road between city1 and city2."""
        if city1 not in self.adj_list:
            self.adj_list[city1] = []
        if city2 not in self.adj_list:
            self.adj_list[city2] = []
        self.adj_list[city1].append(city2)
        self.adj_list[city2].append(city1)

    def has_route(self, start, destination):
        """Checks if a route exists between start and destination using DFS."""
        if start not in self.adj_list or destination not in self.adj_list:
            return False

        visited = set()
        return self._dfs_search(start, destination, visited)

    def _dfs_search(self, current, target, visited):
        if current == target:
            return True

        visited.add(current)
        for neighbor in self.adj_list.get(current, []):
            if neighbor not in visited:
                if self._dfs_search(neighbor, target, visited):
                    return True
        return False

    def find_connected_components(self):
        """Finds all connected components (separate groups of connected cities)."""
        visited = set()
        components = []

        for city in self.adj_list:
            if city not in visited:
                component = []
                self._dfs_collect(city, visited, component)
                components.append(component)
        return components

    def _dfs_collect(self, current, visited, component):
        visited.add(current)
        component.append(current)
        for neighbor in self.adj_list.get(current, []):
            if neighbor not in visited:
                self._dfs_collect(neighbor, visited, component)
```

```python
# Define the network and add roads
network = RoadNetwork()

# Group 1
network.add_road('Seattle', 'Portland')
network.add_road('Portland', 'San Francisco')
network.add_road('San Francisco', 'Los Angeles')

# Group 2
network.add_road('New York', 'Boston')
network.add_road('New York', 'Philadelphia')

# Group 3 (Isolated City)
network.add_road('Honolulu', 'Honolulu')  # Adding a self-loop so it's in the network

# 1. Check routes
city_pairs = [
    ('Seattle', 'Los Angeles'),
    ('Seattle', 'New York'),
    ('Boston', 'Philadelphia')
]

print("--- Route Verification ---")
for start, end in city_pairs:
    exists = network.has_route(start, end)
    print(f"Route from {start} to {end} exists? {'Yes' if exists else 'No'}")

# 2. Find Connected Components
print("\n--- Connected Components of Cities ---")
components = network.find_connected_components()
for idx, component in enumerate(components, 1):
    print(f"Group {idx}: {', '.join(component)}")
```
